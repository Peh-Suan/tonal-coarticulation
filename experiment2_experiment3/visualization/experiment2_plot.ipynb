```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
sns.set_style({"font.family": "serif", "font.serif": ["Times New Roman"]})
palette = sns.color_palette("muted")
colors = {
    "tone55": palette[0],
    "tone35": palette[1],
    "tone51": palette[2],
    "tone33": palette[3],
    "tone21": palette[4],
}
labels = {
    "tone35": "35",
    "tone51": "51",
    "tone55": "55",
    "tone21": "21",
    "tone33": "33",
}

# Load the data
df = pd.read_csv("predicted_probs_man.csv")

fig, ax = plt.subplots(1, 1, figsize=(6, 5), sharey=True)

    
for tone in ['tone51', 'tone55', 'tone35']:
    tone_data = df[df['preceding_tone'] == tone]
    sns.lineplot(x=tone_data['level'], y=tone_data['predicted'], label=labels[tone], color=colors[tone])
    sns.lineplot(x=tone_data['level'], y=tone_data['lower'], linestyle='--', legend=False, color=colors[tone], alpha=.5)
    sns.lineplot(x=tone_data['level'], y=tone_data['upper'], linestyle='--', legend=False, color=colors[tone], alpha=.5)
    
ax.set_xlabel("Level", fontsize=15)
ax.set_ylabel("Predicted falling tone choice probability", fontsize=15)
ax.legend(title="Preceding tone", fontsize=15, title_fontsize=15)
plt.xlim(1, 10)
ax.set_yticklabels([round(i, 1) for i in ax.get_yticks()], size = 15)
ax.set_xticklabels([int(i) for i in ax.get_xticks()], size = 15)

plt.tight_layout()
plt.savefig("logistic_regression_visualization_man.png", dpi=300, bbox_inches=0)
plt.show()
```
